# Importando os Dados do GitHub

Para importar os dados do repositório indicado, utilizamos a URL do arquivo no formato bruto (`raw.githubusercontent.com`) e o carregamos diretamente com o Pandas.

````python
import pandas as pd

# URL do arquivo CSV bruto no GitHub
url = "https://raw.githubusercontent.com/Sainathmudigonda/Student-Result-Analysis/main/student_exam_scores.csv"

# Carregando os dados em um DataFrame
df = pd.read_csv(url)

# Exibindo as primeiras linhas da tabela
display(df.head())
````

````python
# Exibindo informações gerais sobre as colunas e tipos de dados
df.info()
````

# Gráfico de Dispersão: Horas de Estudo vs. Nota da Prova

Vamos plotar um gráfico para visualizar se existe alguma correlação visual entre a quantidade de horas que os alunos estudaram e as notas que eles obtiveram no exame.

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Definindo o estilo do gráfico
sns.set_theme(style="whitegrid")

# Criando a figura
plt.figure(figsize=(10, 6))

# Criando o gráfico de dispersão (scatter plot)
sns.scatterplot(
    data=df,
    x='hours_studied',
    y='exam_score',
    color='blue',
    alpha=0.7,
    edgecolor='w',
    s=80
)

# Adicionando títulos e rótulos aos eixos
plt.title('Relação entre Horas de Estudo e Nota da Prova', fontsize=14, pad=15)
plt.xlabel('Horas de Estudo (hours_studied)', fontsize=12)
plt.ylabel('Nota da Prova (exam_score)', fontsize=12)

# Exibindo o gráfico
plt.show()
````

# Adicionando Regressão Linear ao Gráfico de Dispersão

Vamos utilizar a biblioteca `scikit-learn` para ajustar um modelo de Regressão Linear aos nossos dados de `hours_studied` e `exam_score`. Em seguida, extrairemos a equação da reta ($Y = aX + b$) e o coeficiente de determinação ($R^2$) para exibi-los diretamente no gráfico.

````python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt
import seaborn as sns

# Preparando os dados para o modelo (o scikit-learn espera matrizes 2D para o X)
X = df[['hours_studied']]
y = df['exam_score']

# Ajustando o modelo de regressão linear
model = LinearRegression()
model.fit(X, y)

# Obtendo os coeficientes da regressão
coef_a = model.coef_[0]
intercept_b = model.intercept_

# Fazendo as previsões para traçar a linha de tendência
y_pred = model.predict(X)

# Calculando o R²
r2 = r2_score(y, y_pred)

# Definindo a equação formatada como string
equacao = f"Y = {coef_a:.2f}X + {intercept_b:.2f}"
metricas_texto = f"{equacao}\nR² = {r2:.3f}"

# Criando o gráfico atualizado
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Pontos de dispersão
sns.scatterplot(
    data=df,
    x='hours_studied',
    y='exam_score',
    color='blue',
    alpha=0.6,
    edgecolor='w',
    s=80,
    label='Dados Reais'
)

# Linha de tendência
plt.plot(
    df['hours_studied'],
    y_pred,
    color='red',
    linewidth=2.5,
    label='Linha de Tendência (Regressão Linear)'
)

# Adicionando a equação e o R² no gráfico
# Usamos coordenadas relativas ao gráfico para posicionar o texto de forma limpa
plt.gca().text(
    0.05, 0.90,
    metricas_texto,
    transform=plt.gca().transAxes,
    fontsize=12,
    verticalalignment='top',
    bbox=dict(boxstyle="round,pad=0.5", facecolor="white", edgecolor="gray", alpha=0.9)
)

# Título e rótulos
plt.title('Regressão Linear: Horas de Estudo vs. Nota da Prova', fontsize=14, pad=15)
plt.xlabel('Horas de Estudo (hours_studied)', fontsize=12)
plt.ylabel('Nota da Prova (exam_score)', fontsize=12)
plt.legend(loc='lower right')

# Exibindo o gráfico
plt.show()
````
