```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sakshisatre/ice-cream-sales-dataset")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

# Listar os arquivos no diretório do dataset
print("Arquivos no diretório:", os.listdir(path))
```

```python
# Encontrar o arquivo CSV e carregá-lo em um DataFrame
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
if csv_files:
    csv_path = os.path.join(path, csv_files[0])
    df = pd.read_csv(csv_path)
    print(f"Arquivo '{csv_files[0]}' carregado com sucesso!")
    display(df.head())
else:
    print("Nenhum arquivo CSV encontrado no diretório.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
import numpy as np

# Preparar as variáveis independentes (X) e dependentes (y)
X = df[['Temperature']]
y = df['Revenue']

# Ajustar o modelo de regressão linear
model = LinearRegression()
model.fit(X, y)

# Obter os coeficientes
coeficiente_angular = model.coef_[0]
intercepto = model.intercept_
r_quadrado = model.score(X, y)

print(f"Coeficiente Angular (Inclinação): {coeficiente_angular:.4f}")
print(f"Coeficiente Linear (Intercepto): {intercepto:.4f}")
print(f"Coeficiente de Determinação (R²): {r_quadrado:.4f}")
```

```python
# Gerar pontos para a linha de regressão
X_pred = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
y_pred = model.predict(X_pred)

# Configurar o gráfico
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Temperature', y='Revenue', alpha=0.7, label='Dados Reais')
plt.plot(X_pred, y_pred, color='red', linewidth=2, label=f'Linha de Regressão\ny = {coeficiente_angular:.2f}x + {intercepto:.2f}')

# Adicionar títulos e rótulos
plt.title('Temperatura vs. Faturamento (Vendas de Sorvete)', fontsize=14)
plt.xlabel('Temperatura (°C)', fontsize=12)
plt.ylabel('Faturamento ($)', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)

# Exibir gráfico
plt.show()
```
